## Ejercicio 1 — Inversión con X

Aplicar X a un qubit inicializado en $|0\rangle$ y presentar circuito, vector, ket y esfera de Bloch.

```python
# Importar las herramientas.
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex, plot_bloch_multivector
from IPython.display import display

# Crear el circuito y guardar el estado inicial |0⟩.
circ_x = QuantumCircuit(1)
psi_0 = Statevector(circ_x)

# SOLUCIÓN: aplicar X al qubit 0.
circ_x.x(0)
psi_1 = Statevector(circ_x)

# Mostrar todas las representaciones solicitadas.
display(circ_x.draw('mpl'))

print("👉 Estado inicial:")
display(array_to_latex(psi_0.data))
display(psi_0.draw('latex'))
display(plot_bloch_multivector(psi_0))

print("👉 Estado después de X:")
display(array_to_latex(psi_1.data))
display(psi_1.draw('latex'))
display(plot_bloch_multivector(psi_1))
```

---

#### ✅ ¿A qué estado se transforma `|0⟩` después de aplicar la compuerta X?
El qubit pasa al estado `|1⟩`.

---

#### ✅ ¿Cómo se refleja ese cambio en la esfera de Bloch?
El vector cambia del **polo norte (Z+)** al **polo sur (Z−)**.

---

📌 **Conclusión**:  
La compuerta **X** invierte el estado del qubit.  
`X |0⟩ = |1⟩`  
Visualmente, esto equivale a un giro de 180° sobre el eje X en la esfera de Bloch.

## Ejercicio 2 — Superposición con H

Aplicar H a un qubit inicializado en $|0\rangle$ y presentar las mismas cuatro representaciones.

```python
# Crear un circuito nuevo y guardar el estado inicial.
circ_h = QuantumCircuit(1)
psi_0 = Statevector(circ_h)

# SOLUCIÓN: aplicar H al qubit 0.
circ_h.h(0)
psi_1 = Statevector(circ_h)

# Mostrar todas las representaciones solicitadas.
display(circ_h.draw('mpl'))

print("👉 Estado inicial:")
display(array_to_latex(psi_0.data))
display(psi_0.draw('latex'))
display(plot_bloch_multivector(psi_0))

print("👉 Estado después de H:")
display(array_to_latex(psi_1.data))
display(psi_1.draw('latex'))
display(plot_bloch_multivector(psi_1))
```

---

#### ✅ ¿Qué combinación de estados observas en el vector?
\[
|ψ⟩ = \frac{1}{\sqrt{2}}(|0⟩ + |1⟩)
\]

---

#### ✅ ¿En qué eje se representa el estado dentro de la esfera de Bloch?
Sobre el **eje X positivo**, en el plano ecuatorial.

---

#### ✅ ¿Qué diferencia ves con respecto a la compuerta X?
La X cambia de un estado base a otro (`|0⟩` ↔ `|1⟩`),  
La Hadamard crea una **superposición cuántica**.

---

📌 **Conclusión**:  
La compuerta **Hadamard** transforma un estado base en una combinación lineal de ambos.  
`H |0⟩ = (|0⟩ + |1⟩)/√2`  
Esto inicia el comportamiento cuántico real: **superposición e interferencia**.
