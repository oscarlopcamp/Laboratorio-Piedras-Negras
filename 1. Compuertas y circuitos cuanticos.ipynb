# 🟪 SECCIÓN 01

# 🧱 Construcción de circuitos cuánticos

Antes de aplicar operaciones necesitamos colocar los qubits dentro de una estructura que permita manipularlos: el **circuito cuántico**.

Un circuito representa una secuencia de compuertas, mediciones e inicializaciones. Visualmente se interpreta de la siguiente manera:

| Elemento | Significado |
|---|---|
| Línea horizontal simple | Evolución de un qubit |
| Línea horizontal doble | Información en bits clásicos |
| Caja con una letra | Compuerta cuántica |
| Punto `●` | Qubit de control |
| Símbolo `⊕` | Qubit objetivo de una CNOT |
| Medidor y flecha | Medición y destino clásico |

### Regla fundamental

> ⏱️ El tiempo avanza **de izquierda a derecha**. Las operaciones que aparecen primero a la izquierda se ejecutan primero.

La siguiente imagen resume estos elementos. Consérvala como guía durante el módulo.

![Circuito cuántico](https://quantum.cloud.ibm.com/assets-docs-learning/_next/image?url=%2Flearning%2Fimages%2Fcourses%2Fquantum-computing-in-practice%2Frunning-quantum-circuits%2Fsimplecircuit.avif&w=3840&q=75)


### 🔎 Cómo leer la imagen paso a paso

1. Localiza las líneas de los qubits y comienza en el extremo izquierdo.
2. Recorre cada línea hacia la derecha.
3. Identifica qué compuerta toca a cada qubit.
4. Si una operación une dos líneas, interprétala como una operación conjunta.
5. Observa las mediciones y sigue sus flechas hasta los bits clásicos.

```text
estado inicial  →  compuerta 1  →  compuerta 2  →  medición
```

> 💡 Una línea representa la evolución del qubit; no significa que conserve siempre el valor con el que comenzó.

### Código, diagrama y matrices

Si aplicamos primero H y después X:

| Representación | Orden |
|---|---|
| Código | `circuito.h(0)` y luego `circuito.x(0)` |
| Diagrama | H aparece a la izquierda de X |
| Matrices | XH∣ψ⟩ |

En el producto matricial, la operación más cercana al ket actúa primero. Por eso las matrices parecen escribirse en sentido contrario al código.

Las siguientes compuertas transforman individualmente un qubit.

## ✳️ 3.1 Compuerta H — superposición

La compuerta Hadamard transforma estados de la base computacional en superposiciones:

$$
H=\frac{1}{\sqrt{2}}\begin{pmatrix}1&1\\1&-1\end{pmatrix}
$$

| Entrada | Salida |
|:--:|:--:|
| ∣0⟩ | ∣+⟩ = (∣0⟩ + ∣1⟩)/√2 |
| ∣1⟩ | ∣−⟩ = (∣0⟩−∣1⟩)/√2 |

Al medir cualquiera de estas superposiciones en la base computacional se obtiene 50 % de `0` y 50 % de `1`. El signo distingue sus fases relativas.

````python
# Importar las herramientas necesarias para este ejemplo.
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex, plot_bloch_multivector
from IPython.display import display

# PASO 1: crear un circuito de un qubit.
# Todos los qubits de Qiskit comienzan automáticamente en el estado |0⟩.
circ_h = QuantumCircuit(1)

# PASO 2: guardar una fotografía matemática del estado inicial.
psi_0 = Statevector(circ_h)

# PASO 3: aplicar H al qubit 0, que es el primer qubit.
circ_h.h(0)

# PASO 4: guardar el nuevo estado después de aplicar H.
psi_1 = Statevector(circ_h)

# PASO 5: mostrar el circuito.
display(circ_h.draw('mpl'))

# Mostrar el estado inicial como vector y como ket.
print("👉 Estado inicial:")
print("➤ Vector:")
display(array_to_latex(psi_0.data))  # .data contiene las amplitudes.
print("➤ Ket:")
display(psi_0.draw('latex'))

# \n agrega una línea en blanco antes del siguiente mensaje.
print("\n👉 Estado después de aplicar Hadamard:")
print("➤ Vector:")
display(array_to_latex(psi_1.data))
print("➤ Ket:")
display(psi_1.draw('latex'))
````

````python
# Esta celda utiliza psi_0 y psi_1, creados en la celda anterior.
# Por eso debemos ejecutar las celdas en orden, de arriba hacia abajo.

print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después de aplicar Hadamard:")
display(plot_bloch_multivector(psi_1))
````

> 📌 **Idea clave:** H saca al qubit de la base clásica y permite producir superposición e interferencia. Además, dos H consecutivas se cancelan: $H^2=I$.

## ❎ 3.2 Compuerta X — inversión del qubit

La compuerta X es el equivalente cuántico de una negación:

$$
X=\begin{pmatrix}0&1\\1&0\end{pmatrix}
$$

| Entrada | Salida |
|:--:|:--:|
| ∣0⟩ | ∣1⟩ |
| ∣1⟩ | ∣0⟩ |

Aplicarla dos veces devuelve el estado inicial: $X^2=I$.

````python
from qiskit import QuantumCircuit
from IPython.display import display

# Crear un circuito de un qubit y guardarlo con el nombre circ1.
circ1 = QuantumCircuit(1)

# Aplicar X al primer qubit. El índice del primer qubit es 0.
circ1.x(0)

# Dibujar y mostrar el circuito completo.
display(circ1.draw('mpl'))
````

### Vector de estado y notación ket

`Statevector` permite guardar una fotografía matemática del estado antes y después de una operación.

| Expresión | Información |
|---|---|
| `psi.data` | Vector de amplitudes |
| `array_to_latex(psi.data)` | Vector con formato matemático |
| `psi.draw('latex')` | Estado en notación ket |

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex
from IPython.display import display

# Crear un circuito nuevo. El qubit comienza en |0⟩.
circ1 = QuantumCircuit(1)

# Guardar el estado antes de realizar cualquier operación.
psi_0 = Statevector(circ1)

# Aplicar X al qubit 0.
circ1.x(0)

# Guardar el estado después de X.
psi_1 = Statevector(circ1)

# Mostrar el circuito y después ambos estados.
display(circ1.draw('mpl'))

print("👉 Estado inicial:")
print("➤ En forma de vector:")
display(array_to_latex(psi_0.data))
print("➤ En forma de ket:")
display(psi_0.draw('latex'))

print("\n👉 Estado después de X:")
print("➤ En forma de vector:")
display(array_to_latex(psi_1.data))
print("➤ En forma de ket:")
display(psi_1.draw('latex'))
````

### 🔍 Visualización en la Esfera de Bloch

La siguiente visualización muestra el estado del qubit **antes** y **después** de aplicar la compuerta **X** en la esfera de Bloch.

Recordemos que:

- El estado inicial `|0⟩` se representa en el polo norte de la esfera.
- La compuerta **X** transforma `|0⟩` en `|1⟩`, que se ubica en el polo sur de la esfera.

Con la función `plot_bloch_multivector()` podemos visualizar un qubit, indicando el vector de estado.


````python
# Crear directamente el estado |0⟩ usando su etiqueta "0".
psi_0 = Statevector.from_label('0')

# Mostrar la posición de |0⟩ en la esfera de Bloch.
print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

# Crear un circuito, aplicar X y calcular el estado resultante.
qc = QuantumCircuit(1)
qc.x(0)
psi_1 = Statevector.from_instruction(qc)

# Mostrar la nueva posición después de X.
print("🔵 Estado después de aplicar X:")
display(plot_bloch_multivector(psi_1))
````

## 🟨 3.3 Compuerta Y — inversión con fase

Y también intercambia $|0\rangle$ y $|1\rangle$, pero introduce el número imaginario $i$:

$$
Y=\begin{pmatrix}0&-i\\i&0\end{pmatrix}
$$

| Entrada | Salida |
|:--:|:--:|
| ∣0⟩ | i∣1⟩ |
| ∣1⟩ | −i∣0⟩ |

Al medir inmediatamente, X y Y pueden parecer iguales. La diferencia está en la fase y puede revelarse al combinar Y con otras compuertas.

````python
from qiskit import QuantumCircuit
from IPython.display import display

# Crear un circuito con un qubit.
circ_y = QuantumCircuit(1)

# Aplicar Y al primer qubit.
circ_y.y(0)

# Mostrar dónde aparece la compuerta Y en el circuito.
display(circ_y.draw('mpl'))
````

### Obtener el vector de estado con Qiskit

Con la clase `Statevector` de Qiskit, se puede obtener el estado de un circuito en el punto que se desee:


````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex
from IPython.display import display

# Crear un circuito nuevo y guardar su estado inicial |0⟩.
circ_y = QuantumCircuit(1)
psi_y_0 = Statevector(circ_y)

# Aplicar Y y guardar el estado resultante.
circ_y.y(0)
psi_y_1 = Statevector(circ_y)

# Mostrar el circuito.
display(circ_y.draw('mpl'))

# Comparar el estado anterior y el posterior.
print("👉 Estado inicial:")
display(array_to_latex(psi_y_0.data))
display(psi_y_0.draw('latex'))

print("\n👉 Estado después de Y:")
display(array_to_latex(psi_y_1.data))
display(psi_y_1.draw('latex'))
````

### 🔍 Visualización en la Esfera de Bloch

La siguiente visualización muestra el estado del qubit **antes** y **después** de aplicar la compuerta **Y** en la esfera de Bloch.

Con la función `plot_bloch_multivector()` podemos visualizar el estado cuántico de forma gráfica:


````python
from qiskit.visualization import plot_bloch_multivector
from IPython.display import display

# Preparar y mostrar el estado inicial |0⟩.
psi_y_0 = Statevector.from_label('0')
print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_y_0))

# Construir el circuito Y y calcular su estado final.
qc_y = QuantumCircuit(1)
qc_y.y(0)
psi_y_1 = Statevector.from_instruction(qc_y)

# Mostrar el estado después de aplicar Y.
print("🔵 Estado después de aplicar Y:")
display(plot_bloch_multivector(psi_y_1))
````

> 📌 **¿Por qué X e Y producen la misma esfera al partir de $|0\rangle$?**  
> X produce $|1\rangle$ y Y produce $i|1\rangle$. El factor $i$ multiplica al estado completo y constituye una **fase global**, que no cambia el punto físico en la esfera de Bloch. En una superposición, Y sí puede modificar una fase relativa y orientar el vector de manera diferente.

## 💤 3.4 Compuerta Z — cambio de fase

Z deja intacto $|0\rangle$ y cambia el signo de $|1\rangle$:

$$
Z=\begin{pmatrix}1&0\\0&-1\end{pmatrix}
$$

| Entrada | Salida |
|:--:|:--:|
| ∣0⟩ | ∣0⟩ |
| ∣1⟩ | −∣1⟩ |
| ∣+⟩ | ∣−⟩ |

Z no cambia las probabilidades de `0` y `1`, pero puede cambiar una fase relativa. Esa diferencia es esencial para producir interferencia.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex, plot_bloch_multivector
from IPython.display import display

# PASO 1: crear un circuito y usar X para preparar |1⟩.
circ3 = QuantumCircuit(1)
circ3.x(0)

# Guardar el estado justo antes de aplicar Z.
psi_0 = Statevector(circ3)

# PASO 2: aplicar Z y guardar nuevamente el estado.
circ3.z(0)
psi_1 = Statevector(circ3)

# PASO 3: mostrar el circuito.
display(circ3.draw('mpl'))

# PASO 4: comparar vector, ket y esfera antes y después.
print("👉 Estado antes de Z:")
display(array_to_latex(psi_0.data))
display(psi_0.draw('latex'))
display(plot_bloch_multivector(psi_0))

print("👉 Estado después de Z:")
display(array_to_latex(psi_1.data))
display(psi_1.draw('latex'))
display(plot_bloch_multivector(psi_1))
````

> 📌 Sobre $|1\rangle$, Z produce $-|1\rangle$. El signo afecta a todo el estado y por eso es una fase global: el vector y la esfera de Bloch representan el mismo estado físico.

### Cómo hacer visible el efecto de Z

Para observar el cambio de fase necesitamos una superposición:

1. H transforma $|0\rangle$ en $|+\rangle$.
2. Z transforma $|+\rangle$ en $|-\rangle$.
3. En la esfera de Bloch el vector se desplaza del eje X positivo al eje X negativo.

El siguiente ejemplo conserva el estilo paso a paso y muestra ambos estados por separado.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex, plot_bloch_multivector
from IPython.display import display

# Crear un circuito de un qubit.
circ_z = QuantumCircuit(1)

# H prepara el estado |+⟩ a partir de |0⟩.
circ_z.h(0)
psi_antes_z = Statevector(circ_z)

# Z transforma |+⟩ en |−⟩.
circ_z.z(0)
psi_despues_z = Statevector(circ_z)

# Mostrar el orden H → Z en el circuito.
print("👉 Circuito H → Z:")
display(circ_z.draw('mpl'))

# Observar el estado antes de Z de tres maneras.
print("👉 Estado antes de Z:")
display(array_to_latex(psi_antes_z.data))
display(psi_antes_z.draw('latex'))
display(plot_bloch_multivector(psi_antes_z))

# Observar el estado después de Z de tres maneras.
print("👉 Estado después de Z:")
display(array_to_latex(psi_despues_z.data))
display(psi_despues_z.draw('latex'))
display(plot_bloch_multivector(psi_despues_z))
````

### ✅ Resumen de las compuertas de un qubit

| Compuerta | Transformación principal | Efecto geométrico |
|:--:|---|---|
| **X** | ∣0⟩↔∣1⟩ | Giro de 180° alrededor de X |
| **Y** | Intercambio con fase imaginaria | Giro de 180° alrededor de Y |
| **Z** | Cambia la fase de ∣1⟩ | Giro de 180° alrededor de Z |
| **H** | Intercambia las bases Z y X | Lleva estados base a superposición |

> 🧠 Probabilidades iguales no garantizan que dos estados sean iguales; también debemos observar sus fases.

# 🟧 SECCIÓN 04

# 🔗 Compuerta CNOT (CX)

La compuerta **CNOT** actúa sobre dos qubits al mismo tiempo. También se conoce como **CX**, porque aplica una X de forma controlada.

- El primer qubit es el **control**.
- El segundo qubit es el **objetivo** o *target*.
- El control nunca cambia por la acción de CNOT.
- El objetivo cambia solamente cuando el control vale `1`.

### Símbolos en el circuito

| Símbolo | Significado |
|:--:|---|
| `●` | Qubit de control |
| `⊕` | Qubit objetivo donde puede aplicarse X |
| Línea vertical | Conexión entre control y objetivo |

### Tabla de verdad

| Control | Objetivo antes | Control después | Objetivo después |
|:--:|:--:|:--:|:--:|
| 0 | 0 | 0 | 0 |
| 0 | 1 | 0 | 1 |
| 1 | 0 | 1 | 1 |
| 1 | 1 | 1 | 0 |

En Qiskit escribimos:

```python
circ.cx(control, objetivo)
```

Por ejemplo, `circ.cx(0, 1)` utiliza `q₀` como control y `q₁` como objetivo.

### Lectura con el orden de bits de Qiskit

Qiskit muestra los estados de dos qubits como $q_1q_0$. Para `cx(0, 1)`:

| Entrada | Salida | Explicación |
|:--:|:--:|---|
| ∣00⟩ | ∣00⟩ | `q₀=0`; no cambia `q₁` |
| ∣01⟩ | ∣11⟩ | `q₀=1`; se invierte `q₁` |
| ∣10⟩ | ∣10⟩ | `q₀=0`; no cambia `q₁` |
| ∣11⟩ | ∣01⟩ | `q₀=1`; se invierte `q₁` |

> 🧠 **Regla para recordar:** CNOT pregunta al control “¿eres 1?”. Si la respuesta es sí, invierte el objetivo.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex
from IPython.display import display

# Crear un circuito con dos qubits: q0 y q1.
circ_cnot = QuantumCircuit(2)

# Preparar q0 en |1⟩. Este qubit será el control.
circ_cnot.x(0)
psi_cnot_antes = Statevector(circ_cnot)

# Aplicar CNOT: 0 indica el control y 1 indica el objetivo.
circ_cnot.cx(0, 1)
psi_cnot_despues = Statevector(circ_cnot)

# Mostrar el circuito y comparar los estados.
print("👉 Circuito CNOT:")
display(circ_cnot.draw('mpl'))

print("👉 Estado antes de CNOT:")
display(array_to_latex(psi_cnot_antes.data))
display(psi_cnot_antes.draw('latex'))

print("👉 Estado después de CNOT:")
display(array_to_latex(psi_cnot_despues.data))
display(psi_cnot_despues.draw('latex'))
````

## 4.2 H + CNOT: creación de entrelazamiento

Cuando el control está en un estado clásico, CNOT se comporta según la tabla anterior. Si primero colocamos el control en superposición con H y después aplicamos CNOT, los qubits pueden quedar **entrelazados**.

$$
|\Phi^+\rangle=\frac{|00\rangle+|11\rangle}{\sqrt{2}}
$$

Esto significa que:

| Resultado | Probabilidad |
|:--:|:--:|
| `00` | 50 % |
| `11` | 50 % |
| `01` | 0 % |
| `10` | 0 % |

Los dos qubits quedan correlacionados: al medir aparecen `00` o `11`, pero no `01` ni `10`.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex, plot_bloch_multivector
from IPython.display import display

# Crear un circuito con dos qubits, ambos inicialmente en |0⟩.
circ_bell = QuantumCircuit(2)

# H coloca q0 en superposición.
circ_bell.h(0)

# CNOT relaciona el estado de q1 con el de q0.
circ_bell.cx(0, 1)

# Calcular el estado final conjunto de los dos qubits.
psi_bell = Statevector(circ_bell)

# Mostrar cada representación por separado.
print("👉 Circuito de Bell:")
display(circ_bell.draw('mpl'))

print("👉 Vector:")
display(array_to_latex(psi_bell.data))

print("👉 Ket:")
display(psi_bell.draw('latex'))

print("👉 Esferas individuales:")
display(plot_bloch_multivector(psi_bell))
````

> ⚠️ En las etiquetas de Qiskit, un estado de dos qubits suele mostrarse como $q_1q_0$. Además, en un estado de Bell las esferas individuales aparecen en el centro porque cada qubit por separado no posee un estado puro definido; la información está en la correlación entre ambos.

# 🟥 SECCIÓN 05

# 🔀 El orden de las compuertas

Un circuito se lee de izquierda a derecha y Qiskit coloca las compuertas en el orden en que aparecen en el código.

| Secuencia temporal | Código | Expresión matricial |
|---|---|---|
| H → X | `.h(0)` y después `.x(0)` | XH∣ψ⟩ |
| X → H | `.x(0)` y después `.h(0)` | HX∣ψ⟩ |

En general, $XH\neq HX$. Por eso cambiar el orden puede cambiar el estado final.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex, plot_bloch_multivector
from IPython.display import display

# CIRCUITO A: agregar primero H y después X.
circ_hx = QuantumCircuit(1)
circ_hx.h(0)
circ_hx.x(0)
psi_hx = Statevector(circ_hx)

# CIRCUITO B: usar las mismas compuertas en el orden contrario.
circ_xh = QuantumCircuit(1)
circ_xh.x(0)
circ_xh.h(0)
psi_xh = Statevector(circ_xh)

# Mostrar todas las representaciones del circuito A.
print("👉 Circuito A: H → X")
display(circ_hx.draw('mpl'))
display(array_to_latex(psi_hx.data))
display(psi_hx.draw('latex'))
display(plot_bloch_multivector(psi_hx))

# Mostrar todas las representaciones del circuito B.
print("👉 Circuito B: X → H")
display(circ_xh.draw('mpl'))
display(array_to_latex(psi_xh.data))
display(psi_xh.draw('latex'))
display(plot_bloch_multivector(psi_xh))
````

### ¿Qué debemos observar?

| Circuito | Estado final | Probabilidad de medir 0/1 |
|:--:|:--:|:--:|
| H → X | ∣+⟩ | 50 % / 50 % |
| X → H | ∣−⟩ | 50 % / 50 % |

Las probabilidades directas coinciden, pero la fase relativa es distinta. Agregaremos una H final a ambos circuitos para convertir esa diferencia en un resultado observable.

````python
# Agregar la misma H al final de cada circuito anterior.
# copy() no es necesario aquí porque queremos continuar esos circuitos.
circ_hx.h(0)
psi_hxh = Statevector(circ_hx)

circ_xh.h(0)
psi_xhh = Statevector(circ_xh)

# En el primer circuito, la H final convierte la fase en |0⟩.
print("👉 H → X → H produce:")
display(circ_hx.draw('mpl'))
display(array_to_latex(psi_hxh.data))
display(psi_hxh.draw('latex'))

# En el segundo circuito, la misma H final produce |1⟩.
print("👉 X → H → H produce:")
display(circ_xh.draw('mpl'))
display(array_to_latex(psi_xhh.data))
display(psi_xhh.draw('latex'))
````

### ✅ Regla práctica

| Situación | ¿Importa el orden? |
|---|:--:|
| Dos compuertas actúan sobre qubits distintos | Generalmente no |
| Dos compuertas iguales, como X·X o H·H | Se cancelan |
| Las matrices cumplen $AB=BA$ | No |
| Las matrices cumplen $AB\neq BA$ | Sí, en general |

> 🧠 No basta con identificar qué compuertas aparecen: también debemos saber sobre qué qubit actúan y en qué orden.
