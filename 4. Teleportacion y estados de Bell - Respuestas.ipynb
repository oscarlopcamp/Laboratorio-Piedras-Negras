## Ejercicio 3 — Respuesta

- `Mensaje`: contiene el estado que Alice quiere enviar.
- `Alice`: es la mitad del par entrelazado que conserva Alice.
- `Bob`: es la mitad que conserva Bob y donde aparece el estado final.
- Los bits clásicos: informan a Bob si debe aplicar X, Z, ambas o ninguna.

## Ejercicio 4 — Respuesta

Sin X, el mensaje comienza en `|0⟩`. Bob debe obtener siempre `0`. Los bits de Alice todavía pueden ser `00`, `01`, `10` o `11`.

```python
# Crear registros con nombres.
mensaje = QuantumRegister(1, 'Mensaje')
qubit_alice = QuantumRegister(1, 'Alice')
qubit_bob = QuantumRegister(1, 'Bob')
bits_alice = ClassicalRegister(2, 'bits_Alice')
bit_bob = ClassicalRegister(1, 'bit_Bob')

teleportar_cero = QuantumCircuit(
    mensaje, qubit_alice, qubit_bob, bits_alice, bit_bob
)

# No aplicamos X: Mensaje permanece en |0⟩.
teleportar_cero.h(qubit_alice[0])
teleportar_cero.cx(qubit_alice[0], qubit_bob[0])
teleportar_cero.cx(mensaje[0], qubit_alice[0])
teleportar_cero.h(mensaje[0])
teleportar_cero.measure(mensaje[0], bits_alice[0])
teleportar_cero.measure(qubit_alice[0], bits_alice[1])

with teleportar_cero.if_test((bits_alice[1], 1)):
    teleportar_cero.x(qubit_bob[0])

with teleportar_cero.if_test((bits_alice[0], 1)):
    teleportar_cero.z(qubit_bob[0])

teleportar_cero.measure(qubit_bob[0], bit_bob[0])

preparado = transpile(teleportar_cero, simulador)
conteos = simulador.run(preparado, shots=1000, seed_simulator=7).result().get_counts()

print(conteos)
display(teleportar_cero.draw('mpl'))
display(plot_histogram(conteos))
```

Alice combina y mide sus qubits. Después envía dos bits normales. Bob utiliza esos bits para elegir las correcciones que reconstruyen el estado.
